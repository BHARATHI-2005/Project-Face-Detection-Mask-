```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
import cv2
import numpy as np
import os
import datetime
import pygame
import warnings

warnings.filterwarnings("ignore", category=UserWarning)

print("🚀 Starting model training...")

datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2,
    rotation_range=20,
    zoom_range=0.15,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.15,
    horizontal_flip=True
)

train_gen = datagen.flow_from_directory(
    'C:/Users/hp5cd/Downloads/Train',
    target_size=(224,224),
    batch_size=32,
    class_mode='binary',
    subset='training'
)

val_gen = datagen.flow_from_directory(
    'C:/Users/hp5cd/Downloads/Train',
    target_size=(224,224),
    batch_size=32,
    class_mode='binary',
    subset='validation'
)

base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224,224,3))
base_model.trainable = False

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dropout(0.5)(x)
predictions = Dense(1, activation='sigmoid')(x)

model = Model(inputs=base_model.input, outputs=predictions)

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()

history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=10
)

plt.plot(history.history['accuracy'], label='train acc')
plt.plot(history.history['val_accuracy'], label='val acc')
plt.legend()
plt.title('Accuracy')
plt.show()
plt.plot(history.history['loss'], label='train loss')
plt.plot(history.history['val_loss'], label='val loss')
plt.legend()
plt.title('Loss')
plt.show()

model.save("mask_detector_mobilenet.keras")
print("✅ Model saved as mask_detector_mobilenet.keras")

print("🚀 Starting real-time detection...")

pygame.mixer.init()
alert_path = r"C:/Users/hp5cd/Downloads/[iSongs.info] 01 - Andhamaa Andhamaa.mp3"

if not os.path.isfile(alert_path):
    print(f"⚠ ALERT FILE MISSING: {alert_path}")
else:
    print(f"✅ Alert file found: {alert_path}")

if not os.path.exists("no_mask_captures"):
    os.makedirs("no_mask_captures")

fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter("session_video.avi", fourcc, 20.0, (640,480))

model = tf.keras.models.load_model("mask_detector_mobilenet.keras")
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

no_mask_count = 0
mask_count = 0
timestamps_list = []
start_time = datetime.datetime.now()

cap = cv2.VideoCapture(0)
if not cap.isOpened():
    print("Cannot open camera")
else:
    print("Camera started. Press 'q' or 'Esc' to stop.")

def preprocess_face(face):
    face = cv2.cvtColor(face, cv2.COLOR_BGR2RGB)
    face = cv2.resize(face, (224, 224))
    face = face / 255.0
    return np.expand_dims(face, axis=0)

try:
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        frame = cv2.resize(frame, (640,480))
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        faces = face_cascade.detectMultiScale(gray, 1.1, 4)

        for (x, y, w, h) in faces:
            face_img = frame[y:y+h, x:x+w]
            if face_img.shape[0] == 0 or face_img.shape[1] == 0:
                continue
            pred = model.predict(preprocess_face(face_img))[0][0]
            label = "Mask" if pred < 0.5 else "No Mask"
            color = (0,255,0) if pred < 0.5 else (0,0,255)

            if pred < 0.5:
                mask_count += 1
            else:
                no_mask_count += 1
                timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
                timestamps_list.append(timestamp)
                filename = f"no_mask_captures/no_mask_{timestamp}.jpg"
                cv2.imwrite(filename, frame)
                with open("no_mask_log.txt", "a") as f:
                    f.write(f"No Mask Detected at {timestamp}\n")
                if not pygame.mixer.music.get_busy() and os.path.isfile(alert_path):
                    pygame.mixer.music.load(alert_path)
                    pygame.mixer.music.play()

            cv2.rectangle(frame, (x, y), (x + w, y + h), color, 2)
            cv2.putText(frame, f"{label} {pred:.2f}", (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.8, color, 2)

        total = mask_count + no_mask_count
        percent_mask = (mask_count / total) * 100 if total > 0 else 0
        percent_no_mask = (no_mask_count / total) * 100 if total > 0 else 0
        cv2.putText(frame, f"Mask: {percent_mask:.1f}% No Mask: {percent_no_mask:.1f}%", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255,255,255), 2)

        cv2.imshow("Mask Detection", frame)
        out.write(frame)

        key = cv2.waitKey(1) & 0xFF
        if key == ord('q') or key == 27:
            break

finally:
    pygame.mixer.music.stop()
    cap.release()
    out.release()
    cv2.destroyAllWindows()

    end_time = datetime.datetime.now()
    with open("session_report.txt", "w") as f:
        f.write("Session Report\n")
        f.write(f"{'='*20}\n")
        f.write(f"Session started: {start_time}\n")
        f.write(f"Session ended: {end_time}\n")
        f.write(f"Total no-mask detections: {no_mask_count}\n")
        f.write(f"Total mask detections: {mask_count}\n")
        f.write("Detections at:\n")
        for ts in timestamps_list:
            f.write(f"  - {ts}\n")

    print("✅ Session report saved as session_report.txt")
    print("✅ Session video saved as session_video.avi")
```
